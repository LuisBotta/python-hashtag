# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
import pyautogui as gui
import time

# pyautogui.click -> clicar com o mouse
# pyautogui.write -> escrever um texto
# pyautogui.press -> aperta uma tecla
# pyautogui.hotkey -> combinação de teclas (crtl c por ex)

gui.PAUSE = 0.5 #configurando pausa de 1 segundo entre os passos da automação

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga" #variavel do link do sistema

gui.press("win") #apertando o botão windows
gui.write("edge") #escrevendo edge na barra de busca
gui.press("enter") #pressionando enter
time.sleep(2) #aguardando 2 segundos para ir ao próximo passo
gui.write(link) #digitando o link na barra de endereço do navegador
gui.press("enter") #pressionando enter
time.sleep(2) #aguardando 2 segundos para ir ao próximo passo
```

```python
#print(gui.position()) #para achar a posição do cursor
time.sleep(2)
gui.click(x=-1553, y=20,clicks=2)
time.sleep(2)
gui.click(x=-115, y=28,clicks=1)
time.sleep(2)
gui.click(x=-361, y=119,clicks=2)
time.sleep(5)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas as pd

arquivo = r"C:\Users\felip\Downloads\Vendas - Dez.xlsx"
tabela = pd.read_excel(arquivo)

#display(tabela)

faturamento = sum(tabela["Valor Final"])
print(faturamento)

qtde_produtos = sum(tabela["Quantidade"])
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
email = "https://mail.google.com"
destinatario = "felipesoad999@hotmail.com"

gui.hotkey("ctrl","t")
gui.write(email)
gui.press("enter")
time.sleep(2)
gui.click(x=-1855, y=-143,clicks=2)
time.sleep(2)
gui.write(destinatario)
gui.press("enter")
time.sleep(2)
gui.click(x=-507, y=217,clicks=1)
gui.write("Fechamento mensal")
gui.click(x=-349, y=353,clicks=2)
gui.write(f"""Boa noite! 
Segue o fechamento 
Faturamento = R$ {faturamento} 
Quantidade vendida = {qtde_produtos}""")
gui.hotkey("ctrl","enter")


```
